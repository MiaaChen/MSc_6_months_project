# Rerun Pocket Models From Separate Disease: No Missing Neighbour Genes

This notebook reruns the pocket model workflow from `separate_disease`, but filters neighbouring genes so only candidates whose `matrix_gene_id` appears in the disease-specific gene-gene matrix subset are used.

Outputs are written to `Desktop/6-months_project/pocket_model_reruns_no_missing/`.

```python
# Editable notebook parameters

SCRIPT_PATH = "/Users/miasmacbook/Desktop/6-months_project/scrip/rerun_pocket_models_from_separate_disease_no_missing.py"
PROJECT_ROOT = "/Users/miasmacbook/Desktop/6-months_project"

# Example: ["123_gene_related_disease"], ["3_gene_related_disease"], or ["10_gene_related_disease"]
FOLDER_NAMES = ["123_gene_related_disease"]

# Use [] to run every disease in the selected folders.
DISEASE_IDS = []

N_RESTARTS = 10
N_ROUNDS = 5000
MAX_GENES_TO_CHANGE = 9
RANDOM_SEED = 42
STABILITY_THRESHOLD = 0.8

REQUIRE_UNIQUE_GENES = False
WRITE_ROUND_HISTORY = False

DRAW_DISTRIBUTION_GRAPHS = True
DISTRIBUTION_RUNS = 100
DISTRIBUTION_SEED_START = 42
EXTREME_SEARCH_MAX_COMBINATIONS = 100000
EXTREME_SEARCH_RESTARTS = 40
EXTREME_SEARCH_ROUNDS = N_ROUNDS
EXTREME_SEARCH_SEED = 500000

# Leave as None to write to pocket_model_reruns_no_missing/<folder_name>
OUTPUT_DIR = None
RANDOM_SUMMARY_FILE = None
RANDOM_DISTRIBUTION_FILE = None
GLOBAL_SUMMARY_FILE = None
```

## Run

```python
import sys
from pathlib import Path
import runpy

script_path = Path(SCRIPT_PATH)
script_globals = runpy.run_path(str(script_path), run_name="not_main")
main = script_globals["main"]

argv = [
    "rerun_pocket_models_from_separate_disease_no_missing.ipynb",
    "--project-root", str(PROJECT_ROOT),
    "--n-restarts", str(N_RESTARTS),
    "--n-rounds", str(N_ROUNDS),
    "--max-genes-to-change", str(MAX_GENES_TO_CHANGE),
    "--random-seed", str(RANDOM_SEED),
    "--stability-threshold", str(STABILITY_THRESHOLD),
    "--distribution-runs", str(DISTRIBUTION_RUNS),
    "--distribution-seed-start", str(DISTRIBUTION_SEED_START),
    "--extreme-search-max-combinations", str(EXTREME_SEARCH_MAX_COMBINATIONS),
    "--extreme-search-restarts", str(EXTREME_SEARCH_RESTARTS),
    "--extreme-search-rounds", str(EXTREME_SEARCH_ROUNDS),
    "--extreme-search-seed", str(EXTREME_SEARCH_SEED),
]

for folder_name in FOLDER_NAMES:
    argv.extend(["--folder-name", str(folder_name)])

for disease_id in DISEASE_IDS:
    argv.extend(["--disease-id", str(disease_id)])

if REQUIRE_UNIQUE_GENES:
    argv.append("--require-unique-genes")

if WRITE_ROUND_HISTORY:
    argv.append("--write-round-history")

if not DRAW_DISTRIBUTION_GRAPHS:
    argv.append("--skip-distribution-graphs")

if OUTPUT_DIR:
    argv.extend(["--output-dir", str(OUTPUT_DIR)])

if RANDOM_SUMMARY_FILE:
    argv.extend(["--random-summary-file", str(RANDOM_SUMMARY_FILE)])

if RANDOM_DISTRIBUTION_FILE:
    argv.extend(["--random-distribution-file", str(RANDOM_DISTRIBUTION_FILE)])

if GLOBAL_SUMMARY_FILE:
    argv.extend(["--global-summary-file", str(GLOBAL_SUMMARY_FILE)])

sys.argv = argv
print("Running:", " ".join(argv))
main()
```
